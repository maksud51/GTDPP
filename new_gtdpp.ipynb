```python
import re
from nltk.corpus import stopwords
import nltk
nltk.download('stopwords')

# Define the text cleaning function
def clean_text(text):
    text = text.lower()  # Lowercase the text
    text = re.sub(r'[^a-z\s]', '', text)  # Remove punctuation and numbers
    text = ' '.join([word for word in text.split() if word not in stopwords.words('english')])  # Remove stopwords
    return text

# List of all feature columns to clean
text_columns = ['Product Name', 'Material Composition', 'Material Description', 
                'Textural Features', 'Sustainability Features', 'Color', 
                'Design Features', 'Dyeing Process', 'Packaging Type', 
                'Garment Type', 'Product Design', 'Headline/Screen Text', 
                'Production Process Indicators', 'Features']

# Apply the clean_text function to each of these columns
for col in text_columns:
    data[f'cleaned_{col}'] = data[col].apply(clean_text)

# Display the cleaned data to confirm
data[['Product Name', 'cleaned_Product Name', 'Material Composition', 'cleaned_Material Composition']].head()
```

```python
from transformers import BertTokenizer

# Load the BERT tokenizer
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# Combine relevant cleaned columns into a single 'cleaned_text' column if needed
data['cleaned_text'] = data['cleaned_Product Name'] + " " + data['cleaned_Material Composition'] + " " + data['cleaned_Material Description']

# Tokenize the text
data['tokenized_text'] = data['cleaned_text'].apply(lambda x: tokenizer.encode(x, add_special_tokens=True))
```
